```python
import pandas as pd

file_path = r"C:\Users\akars\OneDrive\Desktop\projects\End-to-End Supply Chain Analytics & Late Delivery Prediction\data\DataCoSupplyChainDataset_Cleaned.csv"

df = pd.read_csv(file_path)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print(df.head(5))
```

```python
from sqlalchemy import create_engine

engine = create_engine(
    "mysql+pymysql://root:12345@localhost/supply_chain"
)
print("Connected successfully!")
```

```python
##Load dataframe to MySQL
table_name = "supply_data"
df.to_sql(table_name, engine, if_exists='replace', index=False)
print(f"Data loaded successfully into the '{table_name}' table in MySQL database.")
```
