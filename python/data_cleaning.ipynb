NUMERIC COLUMNS ANALYZE

```python
numeric_cols = df.select_dtypes(include='number').columns
print(numeric_cols)
```

```python
##NO. of negative values in numeric columns
negative_counts = (df[numeric_cols] < 0).sum()

negative_counts = negative_counts[
    negative_counts > 0
].sort_values(ascending=False)

negative_counts
```

```python
##NO. of zero values in numeric columns
zero_counts = (df[numeric_cols] == 0).sum()

zero_counts = zero_counts[
    zero_counts > 0
].sort_values(ascending=False)

zero_counts
```

INVESTIGATION

For investigating of Alphabetical,Duplicated and missing value , We use excel because of its filtering feature

```python
## Correcting Dates of Order and Shipping Date by removing the time component
date_cols = [
    'order date (DateOrders)',
    'shipping date (DateOrders)'
]

for col in date_cols:
    df_cleaned[col] = pd.to_datetime(
        df_cleaned[col],
        errors='coerce'
    ).dt.date
    
    # Converting object to datetime format for consistency
    df_cleaned[col] = pd.to_datetime(
    df_cleaned[col],
    errors='coerce')
```
